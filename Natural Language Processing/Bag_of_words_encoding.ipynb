```python
vocab = {} #dictionary that maps words to their corresponding integers
word_encoding  = 1
def bag_of_words(text):
  global word_encoding #ties it to the global variable
  words = text.lower().split(" ")
  print(words)
  bag = {}  # stores all of the encodings and their frequency
  for word in words:
    if word in vocab:
      encoding = vocab[word]  # get encoding from vocab
    else:
      vocab[word] = word_encoding
      encoding = word_encoding
      word_encoding += 1

    if encoding in bag:
      bag[encoding] += 1#sums up the frequencies
    else:
      bag[encoding] = 1

  return bag

text = "this is a test to see if this test will work is is test a a"
bag = bag_of_words(text)
print(bag)
print(vocab)

positive_review = "I thought the movie was going to be bad but it was actually amazing"
negative_review = "I thought the movie was going to be amazing but it was actually bad"

pos_bag = bag_of_words(positive_review)
neg_bag = bag_of_words(negative_review)

print("Positive:", pos_bag)
print("Negative:", neg_bag)
```

## Downfalls of this encoding method
## How the Bag of Words Model Loses Word Order

The "bag of words" model inherently loses the order of words because of its fundamental design principle: it treats a document (like a sentence or review) as an unordered collection (a "bag") of words, completely disregarding the grammatical structure, word sequence, or contextual relationships between words.

Think of it like this:

1.  **Counting Frequencies:** The primary goal of bag of words is to count how many times each word appears in a document. For example, in the `bag_of_words` function, you're counting the frequency of each unique word (represented by its encoding).
2.  **No Positional Information:** When you convert text into a bag of words, you're essentially just creating a list of words and their counts. Whether a word appeared at the beginning, middle, or end of a sentence, or whether it was followed by a specific other word, is completely ignored.

Consider these two sentences:

*   "The dog bit the man."
*   "The man bit the dog."

In a simple bag-of-words representation, both sentences would likely produce the exact same bag: `{ 'the': 2, 'dog': 1, 'bit': 1, 'man': 1 }` (assuming these words are already in the `vocab`). The crucial difference in meaning, which comes from the word order, is lost. This is why it's called a "bag" – like a bag of marbles, you know which marbles are in it and how many of each, but not their arrangement.
