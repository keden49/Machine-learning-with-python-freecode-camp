This Encoding system aims to assign each word with a unique integer and maintains order of how they appear in a sentence hence solving the bag of words downfalls.

```python
vocab = {}
word_encoding = 1
def one_hot_encoding(text):
  global word_encoding

  words = text.lower().split(" ")
  encoding = []

  for word in words:
    if word in vocab:
      code = vocab[word]
      encoding.append(code)
    else:
      vocab[word] = word_encoding
      encoding.append(word_encoding)
      word_encoding += 1

  return encoding

text = "this is a test to see if this test will work is is test a a"
encoding = one_hot_encoding(text)
print(encoding)
print(vocab)
```

```python
positive_review = "I thought the movie was going to be bad but it was actually amazing"
negative_review = "I thought the movie was going to be amazing but it was actually bad"

pos_bag = one_hot_encoding(positive_review)
neg_bag = one_hot_encoding(negative_review)

print("Positive:", pos_bag)
print("Negative:", neg_bag)
```

### Downfalls of Integer Encoding

Integer encoding has a few potential downfalls:

*   **Arbitrary Ordinality**: By assigning integers to words, you implicitly introduce an ordinal relationship (e.g., word 1 is 'less than' word 2). This might not be true in reality and can mislead models, especially if the integers are assigned arbitrarily.
*   **Lack of Semantic Meaning**: The integers themselves don't carry any information about the word's meaning or its relationship to other words. 'Cat' encoded as 1 and 'dog' as 2 doesn't imply they are similar in any way.
*   **Increased Feature Space for Categorical Data**: If you have many unique words, integer encoding can lead to a very large number of distinct values. While not a problem for the encoding itself, some models might struggle with high cardinality categorical features.
*   **Not suitable for some models**: Models that expect numerical inputs to have meaningful magnitude differences (e.g., linear models, decision trees) can be negatively impacted by arbitrary integer assignments.

### Explanation of 'Increased Feature Space for Categorical Data'

When we talk about 'Increased Feature Space for Categorical Data' with integer encoding, it means that if you have a lot of unique words (or categories), integer encoding will assign a unique number to each of them. This creates a 'feature' (a column in a dataset, for example) with many different possible values. Some machine learning models can struggle to efficiently process or learn from a feature that has a very large number of distinct values, which is what we mean by high cardinality. It makes the data representation very 'wide' or 'sparse', which can impact model performance and training time.
