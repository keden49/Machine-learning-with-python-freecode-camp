```python
import tensorflow_probability as tfp#provides tools for probabilistic modeling, inference, and statistical computation
import tensorflow as tf
```

```python
'''
transition_distribution
70% it would transition to a cold day and 30% it would transition to a sunny day
80% it would transition to a sunny day and 20% it would transition to a cold day
normal distribution represents continuos values within a range
categorical distribution represents discrete values(specific categories)
observable_distribution represents temperatures that lead to hidden states(sunny or cold)
a sunny day has a mean temp of 15 and a standard dev of 10 range of (5-15-25)
a cold day has a mean temp of 0 and a standard dev of 5 range of (-5-0-5)
probs represents the probability of each state
mean and standard dev are followed by dots to accomodate floating values
loc represents the mean and scale represents the standard deviation


'''
tfd = tfp.distributions # creating shortcut to distributions module
initial_distribution = tfd.Categorical(probs=[0.8, 0.2])#80% chance its a sunny and 20% its cold
transition_distribution = tfd.Categorical(probs=[[0.7, 0.3],[0.2,0.8]])
observation_distribution = tfd.Normal(loc=[0., 15.], scale=[5., 10.])
#
```

```python
'''
Building Hidden Markov Model
num_steps represents the number of days that we would like to predict temp
in this case its 7 days in the future
'''
weather_model = tfd.HiddenMarkovModel(
    initial_distribution=initial_distribution,
    transition_distribution = transition_distribution,
    observation_distribution = observation_distribution,
    num_steps = 7
)
```

```python
'''
According the weather model based on our temperatures and distributions
it estimates the 7 days to follow will have the following temperatures
Analogy for standard deviation:
In a normal distribution:
- Approximately 68% of data falls within 1 standard deviation of the mean.
- Approximately 95% of data falls within 2 standard deviations of the mean.
- Approximately 99.7% of data falls within 3 standard deviations of the mean.
-Therefore,  we have values that are outside the defined standard deviation
-

'''

sample_observations = weather_model.sample()

print("Sampled Observations (Temperatures over 7 days):\n")
for i in range(7):
 print(f"Day {i+1}: {sample_observations[i]}")
```

```python
'''
calculating average mean temperature of each day
it considers different outcomes
eg calculation for day 2

Using current initial_distribution (80% Cold, 20% Sunny for Day 1):

Probability of Day 2 being Cold (P(State_2 = Cold)):
This can happen in two ways:
Day 1 was Cold AND transitioned to Cold: P(State_1 = Cold) * P(Cold -> Cold) = 0.8 * 0.7 = 0.56
Day 1 was Sunny AND transitioned to Cold: P(State_1 = Sunny) * P(Sunny -> Cold) = 0.2 * 0.2 = 0.04
P(State_2 = Cold) = 0.56 + 0.04 = 0.60 (60% chance of being cold on Day 2)

Probability of Day 2 being Sunny (P(State_2 = Sunny)):
This can happen in two ways:
Day 1 was Cold AND transitioned to Sunny: P(State_1 = Cold) * P(Cold -> Sunny) = 0.8 * 0.3 = 0.24
Day 1 was Sunny AND transitioned to Sunny: P(State_1 = Sunny) * P(Sunny -> Sunny) = 0.2 * 0.8 = 0.16
P(State_2 = Sunny) = 0.24 + 0.16 = 0.40 (40% chance of being sunny on Day 2)
(As a check, 0.60 + 0.40 = 1.0, so these probabilities are consistent!)

Finally, the Mean Temperature for Day 2:
Mean Temperature for Day 2 = (P(State_2 = Cold) * Mean Cold Temp) + (P(State_2 = Sunny) * Mean Sunny Temp)
= (0.60 * 0) + (0.40 * 15) = 0 + 6.0 = 6.0
'''
mean_day_temp = weather_model.mean()
for i in range(7):
   print(f" Average temperature of Day {i+1}: {mean_day_temp[i]:.2f}")
```

```python
'''
new learned method for looping
enumrate() returns a list of an index together with its corresponding item in the list
the index is in increment order
'''

mean_day_temp = weather_model.mean()
for i,n in enumerate(mean_day_temp,1):
   print(f" Average temperature of Day {i}: {n:.2f}")
```
